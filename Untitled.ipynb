```python
# from unsloth import FastLanguageModel
# import torch
# from datasets import load_dataset
# from trl import SFTTrainer
# from transformers import TrainingArguments

# # 1. Configuration Constants
# max_seq_length = 1024
# dataset_file = "extraction_train.jsonl"
# output_adapter_dir = "phi3-extractor-lora"

# print("Loading pre-trained Phi-3.5-mini-instruct in 4-bit...")
# # 2. Load Base Model with 4-bit Quantization
# model, tokenizer = FastLanguageModel.from_pretrained(
#     model_name = "unsloth/Phi-3.5-mini-instruct",
#     max_seq_length = max_seq_length,
#     load_in_4bit = True,
# )

# # 3. Apply Low-Rank Adaptation (LoRA)
# model = FastLanguageModel.get_peft_model(
#     model,
#     r = 16,
#     target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
#                       "gate_proj", "up_proj", "down_proj"],
#     lora_alpha = 16,
#     lora_dropout = 0,
#     bias = "none",
#     use_gradient_checkpointing = "unsloth", # Saves massive amounts of VRAM
#     random_state = 3407,
# )

# # 4. Define the Dataset Alpaca Prompt Template
# alpaca_prompt = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

# ### Instruction:
# {instruction}

# ### Input:
# {input}

# ### Response:
# {output}"""

# EOS_TOKEN = tokenizer.eos_token

# def formatting_prompts_func(examples):
#     instructions = examples["instruction"]
#     inputs       = examples["input"]
#     outputs      = examples["output"]
#     texts = []
#     for instruction, input, output in zip(instructions, inputs, outputs):
#         # The EOS token is vital so the model knows exactly when to stop generating
#         text = alpaca_prompt.format(instruction=instruction, input=input, output=output) + EOS_TOKEN
#         texts.append(text)
#     return { "text" : texts }

# print("Processing local JSONL dataset...")
# # 5. Load and Map Dataset
# dataset = load_dataset("json", data_files={"train": dataset_file}, split="train")
# dataset = dataset.map(formatting_prompts_func, batched = True)

# # 6. Initialize the Supervised Fine-Tuning (SFT) Trainer
# trainer = SFTTrainer(
#     model = model,
#     tokenizer = tokenizer,
#     train_dataset = dataset,
#     dataset_text_field = "text",
#     max_seq_length = max_seq_length,
#     dataset_num_proc = 2,
#     args = TrainingArguments(
#         per_device_train_batch_size = 2,
#         gradient_accumulation_steps = 4,
#         warmup_steps = 5,
#         num_train_epochs = 3, # 3 passes over your 500-sample dataset is optimal
#         learning_rate = 2e-4,
#         fp16 = not torch.cuda.is_bf16_supported(),
#         bf16 = torch.cuda.is_bf16_supported(),
#         logging_steps = 5,
#         optim = "adamw_8bit", # Saves VRAM by quantizing optimizer states
#         weight_decay = 0.01,
#         lr_scheduler_type = "linear",
#         seed = 3407,
#         output_dir = "outputs",
#     ),
# )

# # 7. Execute Training
# print("Starting training pass...")
# trainer_stats = trainer.train()
# print(f"Training complete! Final Loss: {trainer_stats.metrics['train_loss']}")

# # 8. Save the PEFT/LoRA Adapters
# model.save_pretrained(output_adapter_dir)
# tokenizer.save_pretrained(output_adapter_dir)
# print(f"Adapters successfully saved locally to the folder: '{output_adapter_dir}'")


from unsloth import FastLanguageModel
import torch
from datasets import load_dataset
from trl import SFTTrainer, SFTConfig

# 1. Configuration Constants
max_seq_length = 1024
dataset_file = "extraction_train.jsonl"
output_adapter_dir = "phi3-extractor-lora"

print("Loading pre-trained Phi-3.5-mini-instruct in 4-bit...")
# 2. Load Base Model with 4-bit Quantization
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Phi-3.5-mini-instruct",
    max_seq_length = max_seq_length,
    load_in_4bit = True,
)

# 3. Apply Low-Rank Adaptation (LoRA)
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
)

# 4. Define the Dataset Alpaca Prompt Template
alpaca_prompt = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

### Instruction:
{instruction}

### Input:
{input}

### Response:
{output}"""

EOS_TOKEN = tokenizer.eos_token

def formatting_prompts_func(examples):
    instructions = examples["instruction"]
    inputs       = examples["input"]
    outputs      = examples["output"]
    texts = []
    for instruction, input, output in zip(instructions, inputs, outputs):
        text = alpaca_prompt.format(instruction=instruction, input=input, output=output) + EOS_TOKEN
        texts.append(text)
    return { "text" : texts }

print("Processing local JSONL dataset...")
# 5. Load and Map Dataset
dataset = load_dataset("json", data_files={"train": dataset_file}, split="train")
dataset = dataset.map(formatting_prompts_func, batched = True)

# 6. Initialize the Supervised Fine-Tuning (SFT) Trainer using SFTConfig
trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    args = SFTConfig(
        dataset_text_field = "text",      # <-- Moved into SFTConfig
        max_seq_length = max_seq_length,  # <-- Moved into SFTConfig
        dataset_num_proc = 2,             # <-- Moved into SFTConfig
        per_device_train_batch_size = 2,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,
        num_train_epochs = 3,
        learning_rate = 2e-4,
        fp16 = not torch.cuda.is_bf16_supported(),
        bf16 = torch.cuda.is_bf16_supported(),
        logging_steps = 5,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "outputs",
    ),
)

# 7. Execute Training
print("Starting training pass...")
trainer_stats = trainer.train()
print(f"Training complete! Final Loss: {trainer_stats.metrics['train_loss']}")

# 8. Save the PEFT/LoRA Adapters
model.save_pretrained(output_adapter_dir)
tokenizer.save_pretrained(output_adapter_dir)
print(f"Adapters successfully saved locally to the folder: '{output_adapter_dir}'")
```
